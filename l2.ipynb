```python
import numpy as np 
```

```python
listy = [1,2,3,4,5]
arr = np.array(listy)

print("Array: ",arr)
print("Shape: ",arr.shape)
print("DataType: ",arr.dtype)
```

```python
zm = np.zeros((3,3))
om = np.ones((2,5))

print("\n3x3 Zero Matrix: \n",zm)
print("Shape: ",zm.shape)

print("\n3x3 Ones Matrix: \n",om)
print("Shape: ",om.shape)
```

```python
arrtwelve = np.arange(1,13)
resarr = arrtwelve.reshape(3,4)

print("\nOriginal Array:", arrtwelve)
print("Shape:", arrtwelve.shape)


print("\nReshaped Array:\n", resarr)
print("Shape:", resarr.shape)
```

```python
randarr = np.random.randint(1,100,10)

print("\nRandom Array:", randarr)
print("Min:", randarr.min())
print("Max:", randarr.max())
print("Mean:", randarr.mean())
print("Sum:", randarr.sum())
print("Standard Deviation:", randarr.std())
```
